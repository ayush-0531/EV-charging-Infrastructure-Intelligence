```python
# IQR calculation 
```

```python
q1 = df["Power (kW)"].quantile(0.25)
q3 = df["Power (kW)"].quantile(0.75)
```

```python
iqr = q3-q1
```

```python
print(f"q1 is {q1}")
print(f"q3 is {q3}")
print(f"iqr is {iqr}")
```

```python
lb = q1 - 1.5*iqr
ub = q3 + 1.5*iqr

print(f"lower bound is {lb}")
print(f"upper bound is {ub}")
```

```python
df[(df["Power (kW)"] < lb ) | (df["Power (kW)"] > ub)]
```
